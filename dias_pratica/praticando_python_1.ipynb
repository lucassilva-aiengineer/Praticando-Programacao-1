```python
# Operadores de comparação

def operadores_comparacao():
    a = 1
    b = 10

    if a == b:
        print("Os dois inteiros são iguais!")

    else:
        print("Não são iguais")


    if a > 10:
      print("O número a é maior que o número b")

    elif a < 10:
      print("O número a é menor que 10")

    else:
      print("O número a é igual a 10")

    if a <= b:
      print("O número a é menor ou igual ao número b")

    elif a > b:
      print("O número a é maior ou igual ao número b")

```

```python
def parte_3()-> None:

  """
  and - Retorna a segunda proposição somente quando
  a primeira é verdadeira
  """

  print(0 and 1) ## 0
  print(1 and 2) ## 2
  print(False and 'Hey') # False O and retorna a segunda apenas quanda a primeira
  print([] and True) # [] # proposição possui valor verdade verdadeiro.
  print(True and False) # False
```
